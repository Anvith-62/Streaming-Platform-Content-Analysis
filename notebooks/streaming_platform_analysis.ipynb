# Streaming Platform Analysis Project

This notebook analyzes streaming platform content using Pandas and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/streaming_platform_sample.csv')
df.head()
```

```python
df.info()
df.isnull().sum()
```

## Movies vs TV Shows

```python
df['type'].value_counts().plot(kind='bar')
plt.title('Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```

## Platform-wise Content Count

```python
df['platform'].value_counts().plot(kind='bar')
plt.title('Content Count by Platform')
plt.xlabel('Platform')
plt.ylabel('Count')
plt.show()
```

## Top Genres

```python
genre_df = df.assign(genre=df['listed_in'].str.split(', ')).explode('genre')
genre_df['genre'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```
